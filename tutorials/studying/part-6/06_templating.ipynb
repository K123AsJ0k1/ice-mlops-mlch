## Why use Templating?

Templating is a common way in DevOps, MLOps, and LLMOps to create human- and machine-readable configurations for the following reasons:

- A universal engineering pattern for structuring raw configurations into structured data that can be checked with schema validation for misconfigurations and enables records of execution states for reproducibility and auditing (mature)

- Easy way to decouple configuration from execution code that enables creating separate parameters for different scenarios and simplifies complex nested logic for dynamic workflows (abstracted)

- Native serialization compatibility via dictionaries across programming languages, with simple API ingestion in modern tools, which is used as a cross-tool standard in different ML tools (interoperable)

These make templating the default pattern for local-cloud-HPC integration, improving the user experience by letting developers focus on building workflows while keeping templated pipeline inputs in one place, and enabling sharing of demonstrated templates for various use cases.

We can update the state of the parameter dictionary in the following way:

```python
from icebreaker.misc.dict import update_nested_dict

status_update_dict = {
    'local-cluster-parameters': {
        'clusters': {
            'lt1': {
                'activate': True
            },
            'lt2': {
                'activate': True
            },
            'lt3': {
                'activate': True
            }
        }
    },
    'cloud-cluster-parameters': {
        'clusters': {
            'vm2': {
                'activate': True
            }
        }
    }
}

updated_local_cloud_cluster_parameter_yamls = update_nested_dict(
    target_dict = local_cloud_cluster_parameter_yamls, 
    update_dict = status_update_dict 
)
```

```python
updated_local_cloud_cluster_parameter_yamls['local-cluster-parameters']
```

We can create a unified pipeline input dictionary from [fragments](templates/fragments) that we fill with the example [data analysis details](./templates/details/local_cloud_data_analysis_details.yaml), with specific workflow steps, in the following way:

```python
workflow_steps = [
    'step-1',
    'step-2'
]
```

```python
workflow_swift_parameters = {
    'last-auth-time': 0,
    'auth-url': 'tutorial',
    'pre-auth-token': 'tutorial',
    'pre-auth-url': 'tutorial',
    'auth-version': 'tutorial',
    'project-name': 'tutorial',
    'user-domain-name': 'tutorial',
    'project-domain-name': 'tutorial'
}
```

```python
from icebreaker.setup.fragments import compose_yaml_dict
data_analysis_pipeline_arguments = compose_yaml_dict(
    fragments_folder = '(absolute-path)/ice-mlops-mlch/tutorials/studying/part-6/templates/fragments',
    fragments_prefix = '_fragment.yaml',
    root_fragment = 'parameters',
    details_folder = '(absolute-path)/ice-mlops-mlch/tutorials/studying/part-6/templates/details',
    details_prefix = '_data_analysis_details.yaml',
    env_dict = {},
    dict_inputs = {
        'run': 'local-cloud-data-analysis-run',
        'experiment': 'local-cloud-data-analysis-experiment',
        'storage|swift': workflow_swift_parameters,
        'integration|cluster-yamls': updated_local_cloud_cluster_parameter_yamls,
        'integration|workflow-steps': workflow_steps,
        'processing|step-1|general|swift': workflow_swift_parameters,
        'processing|step-2|general|swift': workflow_swift_parameters
    }, 
    output_folder = '(absolute-path)/ice-mlops-mlch/tutorials/studying/part-6/generated',
    output_name = 'local_cloud_data_analysis_pipeline_input'
)
```

This function works by first creating a dictionary of fragments from this list:

- [parameters](./templates/fragments/parameters_fragment.yaml)
- [storage](./templates/fragments/storage_fragment.yaml)
- [mlflow](./templates/fragments/mlflow_fragment.yaml)
- [data-storage](./templates/fragments/data-storage_fragment.yaml)
- [code-storage](./templates/fragments/code-storage_fragments.yaml)
- [log-storage](./templates/fragments/log-storage_fragments.yaml)
- [integration](./templates/fragments/integration_fragment.yaml)

It then starts to recursively join the fragments from the root fragment of parameters by iterating through its keys to check for ‘{{dict_fragment}}’ and ‘{{value_fragment}}’ placeholders. The ‘{{dict_fragment}}’ placeholder means a fragment with nested dictionaries, and the ‘{{value_fragment}}’ placeholder means a fragment with normal dictionaries. 

When the recursive process replaces all placeholders with available fragments, it outputs a dictionary and saves it as a YAML file. Check the [example YAML](./generated/local_cloud_data_analysis_pipeline_input_1.yaml) file. We can validate the input dictionary with the following: 

```python
from icebreaker.models.pipeline_arguments import PipelineArguments
pydantic_validation = PipelineArguments.model_validate(data_analysis_pipeline_arguments)
```

With this, we can design and iterate on pipeline input dictionaries by creating suitable Pydantic schemas, drafting YAML fragments that pass these checks, building a dictionary to fill those fragments with details, using the presented functions to build the input dictionary, and confirming that the created dictionary passes all schemas. This improves the developer experience by allowing focus on editing pipeline inputs, which we will demonstrate in practice later.
