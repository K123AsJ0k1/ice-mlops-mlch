## How to setup datasets and models for the demonstration?

Assuming you have done the setup described in the [JupyterLab chapter](../part-1/04_jupyterlab.md) and [Pip chapter](../part-5/05_pip.md) with the ICEbreaker package [(1)](#used-material-1), begin by creating a 'datasets' folder either using JupyterLab or by running the following command:

```python
!mkdir datasets
```

Provide your env path to setup your workflow environment in the following way:

```python
from icebreaker.workflow.utility import workflow_setup_enviroment
workflow_setup = workflow_setup_enviroment(
    env_path = '(fill)/.ssh/.env',
    swift_auth_url = 'https://pouta.csc.fi:5001/v3',
    swift_auth_version = '3',
    swift_pre_auth_url = 'https://a3s.fi:443/swift/v1'
)
```

We can store the internal tutorial material in the following way:

```python
from icebreaker.workflow.use import workflow_handle_internal_datasets
internal_total_time = workflow_handle_internal_datasets(
    swift_client = workflow_setup['swift-client'],
    internal_material_parameters = {
        'material-path': '(fill your path)/ice-mlops-mlch/tutorials/studying',
        'material-prefix': 'part-',
        'excluded-material': [],
        'file-types': [
            'md',
            'ipynb',
            'py',
            'yaml',
            'txt',
            'sh',
            'env'
        ],
        'excluded-folders': [
            'images',
            '_venv',
            'pycache',
            '_checkpoints'
        ],
        'repository-name': 'ice-mlops-mlch',
        'storage-folder': 'datasets',
        'header-start': 'studying',
        'data-prefix': 'internal-tutorial-v1',
        'storage-parameters': {
            'bucket-target': 'experiment',
            'bucket-prefix': '(fill with unique prefix)',
            'bucket-user': 'user@example.com'
        }
    }
)
```

You can confirm that the datasets were stored by going to your cPouta dashboard, finding the bucket, going to DATA/SOURCE, and searching for internal-tutorial-v1. We can store the external open-source datasets |[(2)](#used-material-2), [(3)](#used-material-3), [(4)](#used-material-4), [(5)](#used-material-5), [(6)](#used-material-6)| mentioned in the [LLM application development chapter](./04_llm_application_development.md) by using the [YAML file](./external-dataset-details.yaml) with the details in the following way (be aware that this will take a moment):

```python
from icebreaker.misc.yaml import get_validated_yamls
from icebreaker.workflow.use import workflow_handle_external_datasets

external_dataset_details = get_validated_yamls(
    relative_paths = [
        'external-dataset-details.yaml',
    ],
    yaml_validator = None
)

external_total_time = workflow_handle_external_datasets(
    swift_client = workflow_setup['swift-client'],
    external_material_parameters = {
        'hf-token': workflow_setup['hf-token'],
        'storage-folder': 'datasets',
        'dataset-details': external_dataset_details['external-dataset-details']['details'],
        'data-prefix': 'external',
        'storage-parameters': {
            'bucket-target': 'experiment',
            'bucket-prefix': '(fill with unique prefix)',
            'bucket-user': 'user@example.com'
        }
    }
)
```

Once loading is complete, you can find the datasets again in DATA/SOURCE and search for external. The final thing we need to do is create the 'models' folder and the 'fasttext' folder, download the FastText model lid.176.bin [(7)](#used-material-7), put it into the fasttext folder and store it with the following:

```python
from icebreaker.workflow.utility import workflow_store_models

upload_models_time = workflow_store_models(
    storage_client = workflow_setup['swift-client'],
    storage_parameters = {
        'bucket-target': 'experiment',
        'bucket-prefix': '(fill with unique prefix)',
        'bucket-user': 'user@example.com',
        'swift-parameters': workflow_setup['swift-parameters']
    },
    file_parameters = {
        'model-directory': 'models'
    }
)
```

With this, our datasets and models are ready for use in the subsequent stages of the demonstration workflow.
